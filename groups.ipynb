```python
import so
```

## Load the tag data

```python
filenames = ['android_c.csv', 'csharp_c.csv', 'java_c.csv', 'other_c.csv', 'python_c.csv', 'r_c.csv', 'web_c.csv']
names = ['android', 'c#', 'java', 'other', 'python', 'r', 'web']
```

### Plot weekly data

```python
for i, (name, filename) in enumerate(zip(names, filenames)):
    print(f'Processing: {name}')
    
    escaped_name = name.replace('#', r'\#')
    
    df = so.load_data(filename=f'data/csv/tags/{filename}')
    df.info()

    df['loc'] = df['code'].str.count('\n') + 1
    questions_T_resampled, questions_C_resampled = so.ts(df)

    # question volume
    treated = questions_T_resampled[['PostTypeId']].count()
    control = questions_C_resampled[['PostTypeId']].count()
    so.plot_ts(treated, control, 'PostTypeId', fr'Question Volume with $\mathit{{{escaped_name}}}$ Tags', f'tags/groups/{name}.pdf')

    # question views (mean)
    treated = questions_T_resampled[['ViewCount']].mean()
    control = questions_C_resampled[['ViewCount']].mean()
    so.plot_ts(treated, control, 'ViewCount', fr'Question Views with $\mathit{{{escaped_name}}}$ Tags', f'tags/groups/{name}_views.pdf',
               'Mean views')
    
    # question views (median)
    treated = questions_T_resampled[['ViewCount']].median()
    control = questions_C_resampled[['ViewCount']].median()
    so.plot_ts(treated, control, 'ViewCount', fr'Question Views with $\mathit{{{escaped_name}}}$ Tags', f'tags/groups/median/{name}_views.pdf',
               'Median views')
    
    # question score (mean)
    treated = questions_T_resampled[['Score']].mean()
    control = questions_C_resampled[['Score']].mean()
    so.plot_ts(treated, control, 'Score', fr'Question Score with $\mathit{{{escaped_name}}}$ Tags', f'tags/groups/{name}_score.pdf',
               'Mean score')
    
    # question score (median)
    treated = questions_T_resampled[['Score']].median()
    control = questions_C_resampled[['Score']].median()
    so.plot_ts(treated, control, 'Score', fr'Question Score with $\mathit{{{escaped_name}}}$ Tags', f'tags/groups/median/{name}_score.pdf',
               'Median score')
    
    # question length (mean)
    treated = questions_T_resampled[['line_count']].mean()
    control = questions_C_resampled[['line_count']].mean()
    so.plot_ts(treated, control, 'line_count', fr'Question Length with $\mathit{{{escaped_name}}}$ Tags', f'tags/groups/{name}_line_counts.pdf',
               'Mean length')
    
    # question length (median)
    treated = questions_T_resampled[['line_count']].median()
    control = questions_C_resampled[['line_count']].median()
    so.plot_ts(treated, control, 'line_count', fr'Question Length with $\mathit{{{escaped_name}}}$ Tags', f'tags/groups/median/{name}_line_counts.pdf',
               'Median length')
    
    # question code length (mean)
    treated = questions_T_resampled[['loc']].mean()
    control = questions_C_resampled[['loc']].mean()
    so.plot_ts(treated, control, 'loc', fr'Question LOC with $\mathit{{{escaped_name}}}$ Tags', f'tags/groups/{name}_loc.pdf', 'Mean LOC')

    # question code length (median)
    treated = questions_T_resampled[['loc']].median()
    control = questions_C_resampled[['loc']].median()
    so.plot_ts(treated, control, 'loc', fr'Question LOC with $\mathit{{{escaped_name}}}$ Tags', f'tags/groups/median/{name}_loc.pdf', 'Median LOC')

    # question medium difficulty (mean)
    treated = questions_T_resampled[['Medium']].mean()
    control = questions_C_resampled[['Medium']].mean()
    so.plot_ts(treated, control, 'Medium', fr'Medium Question Diff. with $\mathit{{{escaped_name}}}$ Tags', f'tags/groups/{name}_medium.pdf',
               'Mean Medium Diff.')

    # question medium difficulty (median)
    treated = questions_T_resampled[['Medium']].median()
    control = questions_C_resampled[['Medium']].median()
    so.plot_ts(treated, control, 'Medium', fr'Medium Question Diff. with $\mathit{{{escaped_name}}}$ Tags', f'tags/groups/median/{name}_medium.pdf',
               'Median Medium Diff.')
    
    # question hard difficulty (mean)
    treated = questions_T_resampled[['Hard']].mean()
    control = questions_C_resampled[['Hard']].mean()
    so.plot_ts(treated, control, 'Hard', fr'Hard Question Diff. with $\mathit{{{escaped_name}}}$ Tags', f'tags/groups/{name}_hard.pdf',
               'Mean Hard Diff.')

    # question hard difficulty (median)
    treated = questions_T_resampled[['Hard']].median()
    control = questions_C_resampled[['Hard']].median()
    so.plot_ts(treated, control, 'Hard', fr'Hard Question Diff. with $\mathit{{{escaped_name}}}$ Tags', f'tags/groups/median/{name}_hard.pdf',
               'Median Hard Diff.')
```
