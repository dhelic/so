## Load the posting data

```python
df = so.load_data()
so.did_prepare_data(df)
print(df.info())
```

### Select only data before the treatment begin

```python
df = df[df['P'] == 0]
```

```python
df.info()
```

## Run the regressions on the postings data

```python
questions = so.questions(df)
answers = so.answers(df)
```

```python
questions.info()
```

```python
# question length
results = so.parallel_trend('line_count', questions)
print(results.summary())

# question log length
results = so.parallel_trend('log_lc', questions)
print(results.summary())

# question views
results = so.parallel_trend('ViewCount', questions)
print(results.summary())

# question score
results = so.parallel_trend('Score', questions)
print(results.summary())

# answer length
results = so.parallel_trend('line_count', answers)
print(results.summary())

# answer log length
results = so.parallel_trend('log_lc', answers)
print(results.summary())

# answer score
results = so.parallel_trend('Score', answers)
print(results.summary())
```

## Run the regressions on the tag groups data

```python
filenames = ['android_c.csv', 'csharp_c.csv', 'java_c.csv', 'other_c.csv', 'python_c.csv', 'r_c.csv', 'web_c.csv']
names = ['android', 'c#', 'java', 'other', 'python', 'r', 'web']
```

```python
for i, (name, filename) in enumerate(zip(names, filenames)):
    print(f'Processing: {name}')
    
    df = so.load_data(filename=f'data/csv/tags/{filename}')
    so.did_prepare_data(df, 'tags')
    df.info()

    df = df[df['P'] == 0]

    # question length
    results = so.parallel_trend('line_count', df)
    print(results.summary())

    # question log length
    results = so.parallel_trend('log_lc', df)
    print(results.summary())

    # question views
    results = so.parallel_trend('ViewCount', df)
    print(results.summary())
    
    # question score
    results = so.parallel_trend('Score', df)
    print(results.summary())

    # loc
    results = so.parallel_trend('loc', df)
    print(results.summary())

    # log loc
    results = so.parallel_trend('log_loc', df)
    print(results.summary())

    # medium diff.
    results = so.parallel_trend('Medium', df)
    print(results.summary())

    # hard diff.
    results = so.parallel_trend('Hard', df)
    print(results.summary())
```
